```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

pd.set_option('display.max_columns', 20); # show up to 20 columns when printing a dataframe
pd.set_option('display.max_colwidth', 40); # show up to 40 characters when printing a column of a dataframe

df = pd.read_csv("netflix_titles.csv")
print(df.shape) # print the number of rows and columns in the dataframe
print(df.head()) # print the first 5 rows of the dataframe
print(df.columns) # print the column names of the dataframe

df.info() # prints the summary of the dataframe including the number of non-null values and data types of each column

missing = df.isna().sum() # count the number of missing values in each column
missing_pct = missing / len(df) * 100 # calculate the precentage of missing values in each column
pd.DataFrame({'missing': missing, 'missing percentage': missing_pct}).sort_values('missing percentage', ascending=False) # create a dataframe with the number and percentage of missing values in each column
```
