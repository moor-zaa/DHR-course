Suppose five students report hours studied (x) and a grade (y): 

Student i 1 2 3 4 5

x (hours) 1 2 3 4 5

y (grade) 2 4 5 4 5

```python
import numpy as np



# Defining the data
x = np.array([1,2,3,4,5]);
y = np.array([2,4,5,4,5]);

# Calculating the mean of x and y
x_mean, y_mean = np.mean(x), np.mean(y);

# Calculating the sume of squares --> Reasoning: S_xx = sum of squares of x, S_xy = sum of products of deviations of x and y
S_xx = np.sum((x - x_mean)**2);
S_xy = np.sum((x- x_mean)*(y - y_mean));

# each prediction is given by formula y = b_0 + b_1 * x, where the coefficients b_0 and b_1 are calculated as follows:
b_1 = S_xy / S_xx;  # slope
b_0 = y_mean - b_1 * x_mean;  # intercept

print(f"Regression coefficients: b_0 = {b_0}, b_1 = {b_1}")
```

Regression coefficients: b_0 = 2.2, b_1 = 0.6

### TSS (Total Sum of Squares) : total variation of y around its mean, before using x.
TSS = (y1 - y1_bar)^2 + ... + (yn - yn_bar)^2

### RSS (Residual Sum of Squares) : the variation that is leftover after using x.
e_i = y_i - y_hat_i
RSS = e1^2 + ... en^2



### R^2
R^2 = 1 - RSS/TSS

```python
TSS = np.sum((y- y_mean)**2) # total sum of squares
RSS = np.sum((y - (b_0 + b_1 * x))**2) # residual sum of squares
R_squared = 1 - (RSS/TSS) # R-squared value

n = len(x) # number of observations
RSE = np.sqrt(RSS/(n-2)) # residual standard error

print(f"TSS = {TSS}, RSS = {RSS.round(2)}, R_squared = {R_squared.round(2)}, RSE = {RSE.round(2)}")
```
